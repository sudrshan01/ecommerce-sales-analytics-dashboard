# Load Data


```python
import pandas as pd

df = pd.read_excel("../data/Superstore-final.xlsx")

print(df.head())
```

# Dataset Information & Missing values

```python
print(df.shape)

print(df.info())

print(df.describe())
```

```python
print(df.isnull().sum())
```

# Sales & Profit KPIs

```python
total_sales = df['Sales'].sum()

total_profit = df['Profit'].sum()

total_orders = df['Order ID'].nunique()

print(total_sales)
print(total_profit)
print(total_orders)
```

# Region Analysis

```python
region_sales = df.groupby('Region')['Sales'].sum()

print(region_sales)
```

# Category Analysis

```python
category_profit = df.groupby('Category')['Profit'].sum()

print(category_profit)
```

# Top 10 Products

```python
top_products = df.groupby(
    'Product Name'
)['Sales'].sum().sort_values(
    ascending=False
).head(10)

print(top_products)
```

# Monthly Sales Trend

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Month'] = df['Order Date'].dt.to_period('M')

monthly_sales = df.groupby(
    'Month'
)['Sales'].sum()
```

```python
monthly_sales
```

# Visualization
## Monthly Sales Trend

```python
import matplotlib.pyplot as plt

monthly_sales.plot(
    figsize=(12,6)
)

plt.title("Monthly Sales Trend")
plt.show()
```

## Sales by Region

```python
import seaborn as sns
import matplotlib.pyplot as plt

ax = sns.barplot(
    x=region_sales.index,
    y=region_sales.values
)

# Add exact values on bars
for i, value in enumerate(region_sales.values):
    ax.text(
        i,                  # x position
        value,              # y position
        f'{value:,.0f}',    # format with commas
        ha='center',
        va='bottom'
    )

plt.title("Sales by Region")
plt.ylabel("Sales")
plt.show()
```

## Profit by Category

```python
import seaborn as sns
import matplotlib.pyplot as plt

ax = sns.barplot(
    x=category_profit.index,
    y=category_profit.values
)

# Add exact values
for i, value in enumerate(category_profit.values):
    ax.text(
        i,
        value,
        f'{value:,.0f}',
        ha='center',
        va='bottom'
    )

plt.title("Profit by Category")
plt.ylabel("Profit")
plt.xlabel("Category")

plt.show()
```
